```python
import pandas as pd 
import os
import csv
#traverse a directory
```

```python
def getAllData(modelName, modelType, path):
    metrics = ["loss", "min_k", "zlib", "ref-stablelm-base-alpha-3b-v2"]
    res = []
    res.append(modelName)
    res.append(modelType)
    for metric in metrics:
        fullPath = path+"/"+metric+"_results.json"
        all_res = pd.read_json(fullPath)
        res.append(all_res["metrics"]["roc_auc"])
        res.append(all_res["pr_metrics"]["pr_auc"])
        res.append(all_res["metrics"]["bootstrap_roc_auc_mean"])
        res.append(all_res["metrics"]["bootstrap_roc_auc_std"])
    return res
```

```python
dir_path = "../results_new"
allSubdirs = [d for d in os.listdir(dir_path) if os.path.isdir(os.path.join(dir_path, d))]
subdirs = [item for item in allSubdirs if "pythia" in item]
all_res = []
for subdir in subdirs:
    info = subdir.split('_')
    modelName = info[0]+info[1]
    modelType = info[2]
    relativePath = dir_path+"/"+subdir

    subsubdir = next(name for name in os.listdir(relativePath) if os.path.isdir(os.path.join(relativePath, name)))
    fullRelatiePath = relativePath+"/"+subsubdir+"/github_ngram_13_<0.8_truncated"

    res = getAllData(modelName, modelType, fullRelatiePath)
    all_res.append(res)
```

```python
with open('res.csv', 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    for r in all_res:
        writer.writerow(r)
```
