```python
from scipy.stats import spearmanr

df["Model_Family"] = df["Model"].str.extract(r"^([^-]+)")

# Group by Model_Family and compute Spearman correlation between Size and ROC_AUC
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["ROC_AUC"])
        results[model_family] = round(correlation, 4)
print(results)
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["PR_AUC"])
        results[model_family] = round(correlation, 4)
print(results)
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["ROC_AUC.1"])
        results[model_family] = round(correlation, 4)
print(results)
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["PR_AUC.1"])
        results[model_family] = round(correlation, 4)
print(results)
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["ROC_AUC.2"])
        results[model_family] = round(correlation, 4)
print(results)
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["PR_AUC.2"])
        results[model_family] = round(correlation, 4)
print(results)
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["ROC_AUC.3"])
        results[model_family] = round(correlation, 4)
print(results)
results = {}
for model_family, group in df.groupby("Model_Family"):
    if len(group) > 1:
        correlation, _ = spearmanr(group["Size (MB)"], group["PR_AUC.3"])
        results[model_family] = round(correlation, 4)
print(results)
```
