```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV file
sheet_id = "1tEf8J1kbGZEFTOkoYqIjsdE3PkqBA9bX4MdNfMoLhjQ"
tab_gid = "322169126"
full_url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv&gid={tab_gid}"
df = pd.read_csv(full_url, skiprows=1).iloc[:28,:]
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Compute BLEU loss
df["BLEU Loss"] = df.groupby("Model")["BLEU"].transform("max") - df["BLEU"]

# Plot BLEU Score vs. Model Size
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="Size (MB)", y="BLEU", hue="Compression", style="Model", s=100)
plt.xscale("log")
plt.title("BLEU Score vs. Model Size")
plt.xlabel("Model Size (MB)")
plt.ylabel("BLEU Score")
plt.legend()
plt.grid(True)
plt.show()

# Plot BLEU Loss vs. Compression Type
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="Compression", y="BLEU Loss", palette="Set2")
plt.title("BLEU Score Loss vs. Compression Ratio")
plt.xlabel("Compression Type")
plt.ylabel("BLEU Score Loss")
plt.grid(True)
plt.show()

# Plot BLEU Score Distribution Across Models
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="Model", y="BLEU", hue="Compression", palette="coolwarm")
plt.xticks(rotation=45)
plt.title("BLEU Score Distribution Across Models")
plt.xlabel("Model")
plt.ylabel("BLEU Score")
plt.legend()
plt.grid(True)
plt.show()

# Pareto Front: BLEU Score vs. Model Size
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="Size (MB)", y="BLEU", hue="Compression", style="Model", s=100)
plt.xscale("log")
plt.title("Compression-Performance Tradeoff Curve")
plt.xlabel("Model Size (MB)")
plt.ylabel("BLEU Score")
plt.legend()
plt.grid(True)
plt.show()
```
